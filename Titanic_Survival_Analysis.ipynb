Insight 1: Female passengers had a significantly higher survival rate than male passengers, with 74.03% of females surviving compared with 18.91% of males.

Insight 2: Survival rate increased consistently with ticket fare. High-fare passengers had the highest survival rate at 55.93%, compared with 40.28% for medium-fare and 19.81% for low-fare passengers.

Insight 3: Survival rates generally decreased with increasing age. Children had the highest survival rate at 57.97%, while seniors had the lowest at 26.92%.

Insight 4: The overall survival rate was 38.25%, with 340 out of 889 passengers surviving. This indicates that fewer than half of the analyzed passengers survived.

Insight 5: Passenger class and gender were strongly associated with survival. First-class female passengers had the highest survival rate at 96.74%, while third-class male passengers had the lowest at 13.54%.
